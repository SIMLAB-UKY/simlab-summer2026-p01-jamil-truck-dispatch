# [P01] Week 5: Two-Stage Stochastic Multi-Objective Dispatch
**1. Objective:** Extend the deterministic MILP into a two-stage stochastic program accounting for Exp(120) MTBF breakdowns and triangular repair times using Sample Average Approximation (SAA).

**2. Inputs:** `results/sim_detailed.csv`, Week 4 calibration parameters.

**3. Outputs:** `results/pareto_stochastic.png`, comparative frontier analysis.

**4. Implementation:** Moradi Afrapoli et al. (2024) Eng. Opt. - Stochastic SAA formulation.

```python
# WHAT: Import optimization libraries and setup data handling
# INPUT: None
# OUTPUT: None

import pandas as pd
import numpy as np
import pulp
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from pymoo.core.problem import ElementwiseProblem
from pymoo.termination import get_termination
import matplotlib.pyplot as plt

np.random.seed(42) # Ensure reproducible stochastic scenarios
```

### Mathematical Formulation (Two-Stage Stochastic SAA)

**Objective Function:**
Minimize first-stage assignment costs plus the expected second-stage recourse costs (penalties for unmet demand due to breakdowns).

$$\min_{x} \sum_{i=1}^{T} \sum_{j=1}^{S} C_{ij} x_{ij} + \frac{1}{|\Omega|} \sum_{\omega \in \Omega} \sum_{j=1}^{S} \Psi \cdot s_{j\omega}$$

**Constraints:**
* First-Stage (Assignment): $$\sum_{j=1}^{S} x_{ij} = 1 \quad \forall i$$
* Second-Stage (Scenario Demand): $$\sum_{i=1}^{T} P_{i\omega} x_{ij} + s_{j\omega} \geq D_j \quad \forall j, \omega$$

| Symbol | Definition |
| :--- | :--- |
| $\Omega$ | Set of all generated Monte Carlo scenarios (e.g., $N=50$) |
| $P_{i\omega}$ | Effective production capacity of truck $i$ in scenario $\omega$ after breakdowns |
| $s_{j\omega}$ | Shortfall (unmet demand) at shovel $j$ in scenario $\omega$ |
| $\Psi$ | Recourse penalty cost per ton of unmet demand |

```python
# WHAT: Rebuild deterministic parameters and generate 50 breakdown scenarios
# INPUT: results/sim_detailed.csv
# OUTPUT: P_omega (Matrix of truck capacities across all scenarios)

df = pd.read_csv('results/sim_detailed.csv')
base_haul = df['haul_time_min'].mean()

T_count = 10
S_count = 4

# Base Deterministic Parameters (Matching Week 4)
C = np.random.normal(loc=base_haul, scale=2.0, size=(T_count, S_count)).round(2)
P_base = np.array([100]*5 + [240]*5)
W = (C * np.random.uniform(0.8, 1.5, size=(T_count, S_count))).round(2)
# Shovel Demand (D_j): Increased to 1,650 total tons to stress the 1,700-ton fleet capacity
D = np.array([400, 300, 500, 450])

# Generate SAA Scenarios (Second Stage Recourse)
N_scenarios = 50
shift_mins = 12 * 60
P_omega = np.zeros((N_scenarios, T_count))

for w in range(N_scenarios):
    for i in range(T_count):
        # Sample breakdown time (MTBF 120 hrs)
        time_to_fail = np.random.exponential(120 * 60) 
        
        if time_to_fail < shift_mins:
            # Truck breaks down during shift, loses Tri(30, 60, 120) minutes
            repair = np.random.triangular(30, 60, 120)
            active_time = max(0, shift_mins - repair)
        else:
            active_time = shift_mins # No breakdown
            
        # Calculate reduced capacity based on lost operating time
        P_omega[w, i] = P_base[i] * (active_time / shift_mins)

print(f"Generated {N_scenarios} Monte Carlo breakdown scenarios.")
print(f"Example: Truck 0 Base Capacity: {P_base[0]} tons. Expected Stochastic Capacity: {np.mean(P_omega[:, 0]):.2f} tons.")
```

```python
# WHAT: Solve single-objective cost minimization WITH expected recourse penalties
# INPUT: C, P_omega, D
# OUTPUT: Robust truck-to-shovel assignment

prob = pulp.LpProblem("Stochastic_Dispatch_SAA", pulp.LpMinimize)

# First-Stage Variables: Truck assignment (Binary)
x = pulp.LpVariable.dicts("x", (range(T_count), range(S_count)), cat='Binary')

# Second-Stage Variables: Shortfall per scenario (Continuous >= 0)
s = pulp.LpVariable.dicts("s", (range(N_scenarios), range(S_count)), lowBound=0)

# Objective: Minimize Assignment Cost + Expected Shortfall Penalty
penalty_cost = 500 # High penalty for missing mill targets
expected_penalty = (1/N_scenarios) * pulp.lpSum(penalty_cost * s[w][j] for w in range(N_scenarios) for j in range(S_count))
prob += pulp.lpSum(C[i][j] * x[i][j] for i in range(T_count) for j in range(S_count)) + expected_penalty

# First-Stage Constraints: Truck Capacity
for i in range(T_count):
    prob += pulp.lpSum(x[i][j] for j in range(S_count)) == 1

# Second-Stage Constraints: Meet demand under each specific breakdown scenario
for w in range(N_scenarios):
    for j in range(S_count):
        prob += pulp.lpSum(P_omega[w][i] * x[i][j] for i in range(T_count)) + s[w][j] >= D[j]

prob.solve(pulp.PULP_CBC_CMD(msg=False))

print("--- STOCHASTIC SAA BASELINE ---")
print(f"Status: {pulp.LpStatus[prob.status]}")
print(f"Robust Haulage Cost (Inc. Expected Penalties): {pulp.value(prob.objective):.2f}")
```

```python
# WHAT: Generate robust Pareto frontier balancing Robust Cost, Expected Throughput, and Wear
# INPUT: C, P_omega, W, D
# OUTPUT: res_stochastic (pymoo Result Object)

from pymoo.core.problem import ElementwiseProblem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from pymoo.termination import get_termination
import numpy as np

class StochasticDispatchProblem(ElementwiseProblem):
    def __init__(self):
        # n_ieq_constr changed to 0. We now handle shortfalls via Soft Recourse Penalties!
        super().__init__(n_var=T_count, n_obj=3, n_ieq_constr=0, xl=0, xu=S_count-1)

    def _evaluate(self, x, out, *args, **kwargs):
        assignments = np.round(x).astype(int) 
        
        # 1. Base First-Stage metrics
        first_stage_cost = sum(C[i][assignments[i]] for i in range(T_count))
        wear = sum(W[i][assignments[i]] for i in range(T_count))
        
        # 2. Expected Second-Stage metrics (Stochastic Throughput)
        exp_tons_assigned = [sum(np.mean(P_omega[:, i]) for i in range(T_count) if assignments[i] == j) for j in range(S_count)]
        throughput = -sum(exp_tons_assigned) # Negative for maximization
        
        # 3. STOCHASTIC RECOURSE: Apply the 500-point penalty for any expected shortfalls
        expected_penalty = sum(500 * max(0, D[j] - exp_tons_assigned[j]) for j in range(S_count))
        
        # Objective 1 is now the "Robust Cost" (First Stage + Second Stage Penalties)
        robust_cost = first_stage_cost + expected_penalty
            
        out["F"] = [robust_cost, throughput, wear]
        # out["G"] constraint array removed to allow penalty exploration

algorithm = NSGA2(pop_size=100)
termination = get_termination("n_gen", 200)

res_stochastic = minimize(StochasticDispatchProblem(), algorithm, termination, seed=42, verbose=False)
print(f"Found {len(res_stochastic.F)} optimal stochastic Pareto solutions.")
```

```python
# WHAT: Plot 3D Pareto frontier and export image
# INPUT: res_stochastic.F
# OUTPUT: results/pareto_stochastic.png

costs = res_stochastic.F[:, 0]
throughputs = -res_stochastic.F[:, 1] 
wears = res_stochastic.F[:, 2]

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

scatter = ax.scatter(costs, wears, throughputs, c=throughputs, cmap='plasma', s=50, alpha=0.8)

ax.set_title('Robust Pareto Frontier: Cost vs Wear vs Expected Throughput (SAA)')
ax.set_xlabel('First-Stage Haulage Cost (mins)')
ax.set_ylabel('Equipment Wear Factor')
ax.set_zlabel('Expected Throughput (Tons)')
fig.colorbar(scatter, label='Expected Throughput (Tons)', pad=0.1)

plt.tight_layout()
plt.savefig('results/pareto_stochastic.png')
plt.show()
```
